# NUMPY FOR MATHEDA

```python
import numpy as np
from scipy import stats 


jaguar = [10.6, 9.1, 9.5, 9.8, 10.5, 10.4, 9.5, 11, 10.4, 3, 9.8]


panther = [10.1, 11, 9.1, 20, 9.2, 10.8, 9.9, 9.2, 9.1, 9.1]

def calculate_stats(name, data):
    print(f"--- Statistics for {name} ---")
    print(f"Mean:   {np.mean(data):}")
    print(f"Median: {np.median(data):}")
    
    
    mode_result = stats.mode(data, keepdims=True)
    print(f"Mode:   {mode_result.mode[0]} (count: {mode_result.count[0]})")
    print()

calculate_stats("Jaguar", jaguar)
calculate_stats("Panther", panther)
```

# MEASURES OF CENTRAL TENDENCY 

```python
import numpy as np
from scipy import stats
from statistics import multimode

resistance = np.array([
    [10.6, 9.1, 9.3, 9.8, 10.5, 10.4, 9.5, 11, 10.4, 3, 9.8],
    [10.1, 11, 9.1, 20, 9.2, 10.8, 9.9, 9.2, 9.1, 9.1, np.nan]
])

np.shape(resistance)  

T_resistance = resistance.T

np.shape(T_resistance), np.shape(resistance)

np.mean(resistance, axis=0)

np.mean(resistance, axis=1),np.mean(resistance, axis=0)

np.nanmean(resistance, axis=1)

jag_mean = np.nanmean(resistance, axis=1)[0]
pan_mean = np.nanmean(resistance, axis=1)[1]

print(f"Jaguar mean resistance: {jag_mean}")
print(f"Panther mean resistance: {pan_mean}")
```
